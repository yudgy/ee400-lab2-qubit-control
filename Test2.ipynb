```python
#import the drivers for the pulseblaster, windfreak and the lock in amp

from Ametek_5210 import Ametek5210
from pulseblaster_d import pulseblasterd
from windfreak import windfreak

import time
```

```python
#initalize equiptment
pulser = pulseblasterd('pb')
rf = windfreak('windfreak', 'COM5')
lockin = Ametek5210('lockin', 'GPIB0::12::INSTR')
```

```python
#function to determine delay based on lock in amp time constant (import the lock in amp time constant)
def set_delay(lockin_tc):
    timeconst = {0: '1 ms',
    1: '3 ms',
    2: '10 ms',
    3: '30 ms' ,
    4: '100 ms',
    5: '300 ms',
    6: '1 s',
    7: '3 s',
    8: '10 s',
    9: '30 s',
    10: '100 s',
    11: '300 s',
    12: '1 ks',
    13: '3 ks'}

    timeconst_bw = {'1 ms' :0,
    '3 ms': 1,
    '10 ms': 2,
    '30 ms': 3 ,
    '100 ms': 4,
    '300 ms' : 5,
    '1 s' : 6,
    '3 s' : 7,
    '10 s': 8,
    '30 s': 9,
    '100 s': 10,
    '300 s': 11,
    '1 ks' : 12,
    '3 ks' : 13}

    tc_s = {0: 0.001,  #get timeconstant in secs
    1: 0.003,
    2: 0.01,
    3: 0.03 ,
    4: 0.1,
    5: 0.3,
    6: 1,
    7: 3,
    8: 10,
    9: 30,
    10: 100,
    11: 300,
    12: 1000,
    13: 3000}

    val = timeconst_bw[lockin_tc]
    if(type(val) != int):
        return "Error: invalid input!"
    else:
        return 5* tc_s[val]
```

```python
#Turn on the pulseblaster signals
pulser.program_samplePulse() #will set the sample pulse sequence that you wrote
pulser.start() #will start running the pulse sequence on the programmed ports
```

```python
rf.off()   #turn off in case it is still running
```

```python
#set the sensitivity and the time constant of the lock in amp
tc = '1 s' #lock in amp time constant, you will need to wait 5x this time/point
sens = '10 mV' #this should be the smallest value w/o overloading the lock-in
lockin.sensitivity(sens) 
time.sleep(1)
lockin.time_constant(tc) 

#set the delay based on the lockin sensitivity # confirm delay is unit sensitive
delay = set_delay(tc) #from function above, delay is 5xtc
print(delay)

for x in range (0, 10):
    time.sleep(int(delay)) #set using the time constant of the lockin amp later (5*timeconstant)
    val = lockin.magnitude() #read the magnitude of the lock in amp output

    print(val)
```

```python
pulser.stop()   #turn off the pulseblaster outputs
```
