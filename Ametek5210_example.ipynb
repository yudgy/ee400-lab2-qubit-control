## __Example notebook for Ametek Model 5210 using QCoDeS__

### __Imports and connecting to the instrument__

```python
import json
from qcodes.utils import NumpyJSONEncoder

from Ametek_5210 import Ametek5210 #Need to fix this so you dont have to drag driver to this folder anymore
from qcodes.dataset import load_or_create_experiment
```

### __Enabling Connection to the Instrument__

```python
amtk_lockin = Ametek5210('locking_in', 'GPIB0::12::INSTR')
load_or_create_experiment(experiment_name='Ametek_notebook')
```

### __Running Commands__ 

#### Signal Channel Commands 

```python
# "SEN" Command. Set or Get the full-scale sensititvity. 
amtk_lockin.sensitivity('3 mV')

# "G" Command. Set or Get the status and gain of the signal channel gain vernier control
amtk_lockin.vernier_gain()
```

#### Signal Channel Filters

```python
# "FLT" Command. Set or Get the signal-channel filter mode.
amtk_lockin.filter_operator('LB')

# "ATC" Command. Set filter frequency tuning mode to automatic or manual. 
amtk_lockin.ff_tuning_mode('automatic')

# "FF" Command. Set filter frequency (assuming these are valid values). 
#amtk_lockin.filter_frequency(100, 3)

# LF Command. Sets or reads the mode of the line frequency notch filter.
amtk_lockin.frequency_rejection(2)
```

#### Reference Channel

```python
# "IE" Command. Set or Get the channel source to internal or externa..
amtk_lockin.channel_source('EXT')

# "F2F" Command. Set or Get the reference frequence mode.
amtk_lockin.harmonic_mode(0)

# "P" Command. Set or Get the reference channel phase shifter.
#amtk_lockin.reference_phase()
```

```python
# "FRQ" Command. Get only command that causes the lock-in amplifier to respond based on whether it is unlocked/locked and based on whether it is internal
amtk_lockin.frequency()
```

#### Signal Channel Output Filters

```python
# "XDB" Command. Set or Get the slope of the output filters.
amtk_lockin.output_filters(12)

# "TC" Command. Set or Get the time constant
amtk_lockin.time_constant('1 s')

# "DR" Command. Set or Get the instrument dynamic reserve
amtk_lockin.dynamic_reserve('NORM')

# "EX" Command. Set or Get the X-channel output expansion mode.
amtk_lockin.expansion_control('10X')
```

#### Instrument Outputs (all of these commands are reads, none of them are setters)

```python
# "MAG" Command. Get only command that causes the lock-in to respond with the magnitude output.
amtk_lockin.magnitude()
```

```python
# "PHA" Command. Get only command that causes the lock-in to respond with the signal phase in millidegrees
amtk_lockin.signal_phase()
```

```python
# "MP" Command. Get only command that is equivalent to the compound command. MAG;PHA
amtk_lockin.magnitude_and_phase()
```

```python
# "NN" Command. Get only command that causes the lock-in to respond with the noise, expressed as a percentage of the present full scale sensitivity.
amtk_lockin.noise()
```

#### Auxiliary Inputs

```python
# "TRIG" Command. Set or Get the trigger mode of the ADC converter used for converting both the X-channel and Y-channel outputs and the auxiliary ADC  inputs according to the following table.
amtk_lockin.trig()
```

#### Computer Interfaces

```python
# "N" Command. Get command that causes the lock in amplifier to respond with the overload byte.
amtk_lockin.overload_byte()
```
