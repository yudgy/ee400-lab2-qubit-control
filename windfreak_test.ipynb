## Example Notebook for Windfreak MW Generator using QCoDes

#### Imports and connecting to the instrument

```python
import json
from qcodes.utils import NumpyJSONEncoder

from windfreak import windfreak
from qcodes.dataset import load_or_create_experiment
```

```python
# if you cannot find the COM5 connection, make sure it is connected to the computer
import qt3rfsynthcontrol

qt3rfsynthcontrol.discover_devices()
```

### __Enabling Connection to the Instrument__

```python
rf_gen = windfreak('windfreak', 'COM5')

load_or_create_experiment(experiment_name='windfreak_notebook')
```

### __Set Parameters of the Instrument__

#### Set Channel

```python
rf_gen.set_channel(0)
rf_gen.set_channel()
```

#### Set Frequency

```python
#was 2800.0000000
rf_gen.set_freq()
```

```python
rf_gen.set_freq(2800.0000000)
rf_gen.set_freq()
```

#### Set power level

```python
rf_gen.set_power_dBm(-5.000)
rf_gen.set_power_dBm()
```

#### Turn on the RF module signal

```python
rf_gen.on()
```

#### Turn off the RF module signal

```python
rf_gen.off()
```

#### Print the parameters of the RF module

```python
rf_gen.print_readable_snapshot(update=True)
```
