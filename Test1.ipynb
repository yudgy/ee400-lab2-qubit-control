### This notebook is used to test the Pulseblaster, Windfreak and RF Switch



```python
#import the drivers for the pulseblaster and the windfreak
from pulseblaster_d import pulseblasterd
from windfreak import windfreak
from Ametek_5210 import Ametek5210
```

```python
#Create objects for the pulseblaster and the windfreak

pulser = pulseblasterd('pulseblaster')
rf_gen = windfreak('windfreak', 'COM5')
lockin = Ametek5210('lockin', 'GPIB0::12::INSTR' )
```

```python
#Turn on the pulseblaster signals
pulser.program_samplePulse() #will set the sample pulse sequence that you wrote
pulser.start() #will start running the pulse sequence on the programmed ports
```

```python
#Set the Windfreak RF Output - more info about these functions in windfreak.py and windfreak_test.py
rf_gen.set_channel(0)         #set the channel for the output RF Signal for windfreak
rf_gen.set_power_dBm(-5.000)  #set the power level for the output RF signal
rf_gen.set_freq(2200.0000000) #set the frequency of the output RF signal

rf_gen.on()         #start generating windfreak RF signal output
```

```python
pulser.stop()
```
